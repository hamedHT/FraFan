## Urban Heat Island Mapping with XGBoost + SHAP

هدف: پیش‌بینی دمای سطح زمین با داده‌های ماهواره‌ای و تفسیر عوامل مؤثر.

```python
%pip install -q xgboost shap scikit-learn pandas numpy matplotlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from xgboost import XGBRegressor
import shap
print('Ready')
```

```python
np.random.seed(42)
data = pd.DataFrame({
'NDVI':np.random.rand(1000),
'NDWI':np.random.rand(1000),
'NDBI':np.random.rand(1000),
'B04':np.random.rand(1000),
'B08':np.random.rand(1000),
'B11':np.random.rand(1000)
})
data['LST'] = 25 + 12*data.NDBI - 8*data.NDVI + np.random.normal(0,1,1000)
data.head()
```

## آموزش XGBoost Regression

```python
X=data.drop(columns=['LST'])
y=data['LST']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
model=XGBRegressor(n_estimators=300,max_depth=6,learning_rate=0.05,random_state=42)
model.fit(X_train,y_train)
pred=model.predict(X_test)
print('RMSE:',mean_squared_error(y_test,pred)**0.5)
print('MAE:',mean_absolute_error(y_test,pred))
print('R2:',r2_score(y_test,pred))
```

## Explainable AI با SHAP

```python
explainer=shap.TreeExplainer(model)
values=explainer.shap_values(X_test)
shap.summary_plot(values,X_test)
```
