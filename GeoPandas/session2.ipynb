```python
import matplotlib.pyplot as plt
import geopandas as gpd
```

```python
data = gpd.read_file("gis_osm_adminareas_a_07_1.shp")
data
```

```python
buundary = data.boundary
buundary[1]
```

```python
envelope = data.envelope
envelope[1]
```

```python
intersection = data.intersection(data.geometry[0:9])
intersection[1]
```

```python
data.plot(figsize=(10, 10), color='lightblue', edgecolor='black')
```

```python
data.boundary.plot(figsize=(8,6), color='black')
```

```python
fig, ax = plt.subplots(figsize=(8, 6),dpi=200)
data.boundary.plot(ax=ax, color='black')
data.plot(ax=ax, color='lightblue', edgecolor='black') 
data.centroid.plot(ax=ax, color='red', markersize=10)
```

```python
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = 12
fig, ax = plt.subplots(figsize=(8, 6),dpi=200)
data.plot(ax=ax, cmap='OrRd',column='lastchange', legend=True)
ax.set_title('Map of Administrative Areas', fontsize=16)
ax.set_xlabel('Longitude', fontsize=12)
ax.set_ylabel('Latitude', fontsize=12)
plt.show()
```

```python
ax = data.area.plot.area(figsize=(8, 6), color='lightblue')
```

```python
ax = data.area.plot.density(figsize=(8, 6), color='lightblue')
```
